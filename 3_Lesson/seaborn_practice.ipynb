```python
import seaborn as sns
import scipy

sns.get_dataset_names()
```

```python
# Loading datasets
tips = sns.load_dataset('tips')
iris = sns.load_dataset('iris')
titanic = sns.load_dataset('titanic')
planets = sns.load_dataset('planets')
```

```python
# Scatterplot

sns.scatterplot(x="tip", y="total_bill", data=tips, hue="day", size="size")
```

```python
# Histogram

sns.histplot(tips["tip"], kde=True, bins=15)
```

```python
# Distplot

sns.displot(tips["tip"], kde=True, bins=15)
```

```python
# Barplot

sns.barplot(x="sex", y="tip", data=tips)
```

```python
# Boxplot

sns.boxplot(x="day", y="tip", data=tips, hue="sex")
```

```python
# Stripplot

sns.stripplot(x="day", y="tip", data=tips, hue="sex", dodge=True)
```

```python
# Jointplot

sns.jointplot(x="tip", y="total_bill", data=tips, kind="reg")
```

```python
# Pairplot

sns.pairplot(titanic.select_dtypes(['number']), hue="pclass")
```

```python
# Heatmap

sns.heatmap(titanic.select_dtypes(['number']).corr(), annot=True, cmap="coolwarm")
```

```python
# Clustermap

sns.clustermap(iris.drop("species", axis=1))
```
