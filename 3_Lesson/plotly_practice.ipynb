```python
import pandas as pd
import plotly.express as px
import seaborn as sns

iris = sns.load_dataset('iris')
```

```python
# Barplot
sleep = pd.DataFrame({
    "days": ["Mon", "Tue", "Wed"],
    "sleep_hours": [8, 5, 10],
})

fig = px.bar(
    data_frame=sleep,
    x="days", y="sleep_hours",
    title="Hours of Sleep",
    color="days"
)

fig.show()
```

```python
# Histogram

fig = px.histogram(
    data_frame=iris,
    x="petal_length",
    nbins=20
)

fig.show()
```

```python
# Boxplot

fig = px.box(
    data_frame=iris,
    x="species",
    y="sepal_width",
    points="all",
    hover_data=["species"]
)

fig.show()
```

```python
# Scatterplot

fig = px.scatter(
    data_frame=iris,
    x="sepal_length",
    y="petal_length",
    symbol="species",
    color="species",
    trendline="ols"
)

fig.show()
```

```python
# Lineplot

stocks = pd.read_csv('data/stocks.data.csv')

fig = px.line(
    data_frame=stocks,
    x="Date",
    y="MSFT",
)

fig.show()
```
