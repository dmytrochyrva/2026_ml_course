```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
# Basic X/Y plot
x = [2023, 2024, 2025, 2026]
y = [15, 25, 30, 20]
y1= [19, 53, 45, 29]

plt.plot(x, y, marker="o")
plt.plot(x, y1, marker="o")
plt.grid()
plt.xticks(x)
plt.show()
```

```python
# Bar chart
categories = ["Grains", "Fruit", "Vegetables", "Protein", "Dairy", "Sweets"]
values = [4, 3, 2, 5, 3, 1]

plt.bar(categories, values)
plt.title("Daily Consumption")
plt.show()
```

```python
# Pie chart
categories = ["Freshmen", "Sophomores", "Juniors", "Seniors"]
values = [300, 250, 275, 225]
colors = ["red", "orange", "blue", "green"]

plt.pie(values, labels=categories, autopct='%1.1f%%', colors=colors, explode=[0,0,0,0.1])
plt.show()
```

```python
# Scatter graph
x = [0, 1, 1, 2, 3, 4, 5, 6, 7, 7, 8]
y = [55, 60, 65, 62, 68, 70, 75, 78, 82, 85, 87]

plt.scatter(x, y, label="Class A")
plt.title("Test Scores")
plt.xlabel("Hours Studied")
plt.ylabel("Grade")
plt.legend()
plt.grid()
plt.show()
```

```python
# Histograms
scores = np.random.normal(loc=80, scale=10, size=100)

plt.hist(scores, bins=10, edgecolor="black")
plt.title("Exam Scores")
plt.xlabel("Scores")
plt.ylabel("Num of Students")
plt.show()
```

```python
# Subplots
x = np.array([1,2,3,4,5])

figure, axes = plt.subplots(2, 2)

axes[0,0].plot(x, x*2, color="red")
axes[0,1].bar(x, x**2, color="blue")
axes[1,0].plot(x, x**3, color="green")
axes[1,1].scatter(x, x**4, color="orange")

plt.tight_layout()
plt.show()
```

```python
# Integrating pandas
df = pd.read_csv('pokemon.data.csv')
type_counts = df["Type1"].value_counts(ascending=True)

plt.barh(type_counts.index, type_counts.values)
plt.grid()
plt.show()
```
