## Data Visualization
---

Visualization is used:
- To make it easier to find patterns in data
- To present the model and prove that it is working

Visualization libraries:
- Matplotlib
- Seaborn
- Plotly library

Adding New Features:
- New features can be extracted from the existing ones

Feature Extraction Tools:
- Featuretools
- Anaconda
