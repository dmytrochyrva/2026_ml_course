```python
import pandas as pd
```

```python
# Series
data = [100, 234, 43, 141, 99]
series = pd.Series(data, index=['a', 'b', 'c', 'd', 'e'])
print(series)
print(series.loc['c'])
print(series.iloc[3])
```

```python
# Filtering series
filtered = series[series >= 200]
print(filtered)
```

```python
# DataFrames
data = {
    "Name": ["Spongebob", "Patrick", "Squidward"],
    "Age": [30, 35, 50]
}

dataframe = pd.DataFrame(data)
print(dataframe)
print(dataframe.loc[0])

dataframe['Job'] = ["Cook", "N/A", "Cashier"]
print(dataframe)

new_row = pd.DataFrame([{"Name": "Sandy", "Age": 28, "Job": "Engineer"}])
dataframe = pd.concat([dataframe, new_row])
dataframe
```

```python
# Filtering data
df = pd.read_csv('data/adult.data.csv')

reach = df[df['salary'] == ">50K"]
reach.head()

reach_black = df[(df['salary'] == ">50K") & (df['race'] == 'Black')]
reach_black.head()
```

```python
# Aggregation
df = pd.read_csv('data/adult.data.csv')

df.mean(numeric_only=True)
df.sum(numeric_only=True)
df.min(numeric_only=True)
df.count()

# Single Column
df["age"].max()

# Grouping
group = df.groupby('race')
group["hours-per-week"].mean()
```

```python
# Data cleaning
df = pd.read_csv('data/adult.data.csv')

df = df.drop(columns=["education-num", 'fnlwgt', 'capital-loss', 'capital-gain'])

df['salary'] = df['salary'].replace({"<=50K": "poor", ">50K": "rich"})
df
```
