##### **Terms:**
**AI (Artificial Intelligence)** - A program that can sence, reason, act, and adapt (intel).

**ML (Machine Learning)** - A subset of artificial intelligence that uses algorithms and data to let computers learn patterns and make decisions without explicit, hard-coded rules.

**DL (Deep Learning)** - Teaching neural networks with several layers of neurons.
